# Question 03: [Marks 5]

## Explain what a filter (kernel) is in a CNN and how feature maps are generated during the convolution operation. Discuss how different filters can detect edges, textures, wrinkles, and other facial patterns that are useful for classifying a face as Young, Middle, or Old.


# Write Answer 03:

A **filter (kernel)** is a small matrix of numbers (like 3×3) that a CNN learns during training. It works like a tiny pattern detector that slides over the image.

**How a feature map is made:** The filter is placed on a small patch of the image. Each filter number is multiplied with the matching pixel, and all the products are added to give **one number**. The filter then slides (by the stride) and repeats this everywhere. All these numbers together form the **feature map**. If you use 16 filters, you get 16 feature maps.

**Example:** This filter detects vertical edges:

```
1 0 -1
1 0 -1
1 0 -1
```

It gives a big value where the left side is bright and the right side is dark.

**For faces (Young, Middle, Old):**

- **Edge filters** find the face outline, eyes, nose, and mouth lines.
- **Texture filters** find skin smoothness. Smooth skin points to Young.
- **Wrinkle filters** find many thin lines on the forehead and around the eyes. These point to Middle or Old.
- **Brightness/color filters** can notice gray or white hair, which points to Old.

The CNN learns all these filters by itself and combines them in deeper layers to decide the age group.

# Question 05: [ Marks 10 ]

## Explain padding and stride in CNNs in detail. Then explain the convolutional output size formula and what each symbol in the formula means. You do not need to derive the formula. Show how the output height and width change when you use valid padding, same padding, or a larger stride. Also explain why padding is useful when working with cropped face images.


# Write Answer 05:

**Padding** means adding extra zero pixels around the image border. **Valid padding** (P = 0) adds nothing, so the output shrinks. **Same padding** (P = 1 for a 3×3 filter) keeps the output the same size as the input.

**Stride** is how many pixels the filter moves each step. A larger stride gives a smaller output.

Output size formula: `Output = ⌊(N + 2P − F) / S⌋ + 1`

- N = input size, **P** = padding, F = filter size, S = stride, ⌊ ⌋ = round down

Example (N = 64, F = 3):

| Case | P | S | Output |
|---|---|---|---|
| Valid | 0 | 1 | (64 − 3)/1 + 1 = **62×62** |
| Same | 1 | 1 | (64 + 2 − 3)/1 + 1 = **64×64** |
| Larger stride | 1 | 2 | ⌊(64 + 2 − 3)/2⌋ + 1 = **32×32** |

**Why padding helps with cropped faces:** Details like the chin, ears, and forehead wrinkles are near the border. Padding keeps this information and stops the image from shrinking too fast.

# Question 06: [ Marks 10 ]

## Explain the concept of pooling in CNNs. Compare MaxPooling and AveragePooling in terms of operation, advantages, and impact on feature extraction. Discuss how pooling helps in reducing overfitting and computational complexity when training an age-detection CNN on thousands of face images.


# Write Answer 06:

**Pooling** reduces the size of a feature map by summarizing small regions (like 2×2) into one value. It has no learnable weights. For example, a 2×2 pooling with stride 2 turns a 64×64 map into 32×32.

**MaxPooling vs AveragePooling** (example region: `[1 3; 2 4]`)

| | MaxPooling | AveragePooling |
|---|---|---|
| Operation | Takes the **maximum** value. Result = **4** | Takes the **average** value. Result = (1+3+2+4)/4 = **2.5** |
| Advantage | Keeps the strongest features (sharp edges, wrinkles) | Gives smooth output and keeps overall information |
| Impact on features | Highlights whether a feature is present | Blurs details, so strong features can fade |
| Common use | Between conv blocks (most common) | Near the end (global average pooling) |

**How pooling helps in age-detection CNN:**

- **Less computation:** A 2×2 pool removes 75% of the values, so the next layers train faster and use less memory on thousands of face images.
- **Less overfitting:** Fewer values means fewer parameters in the fully connected layer, so the model cannot memorize tiny details of each image.
- **Small shift tolerance:** If a wrinkle moves by a pixel or two (because of pose or cropping), the pooled output stays almost the same.

# Write Answer 07:

1. **Convolution:** The filter slides over the input. At each position, multiply the filter with the patch (element by element) and add the products. This gives one output value.
2. **Input (5×5):**
   - Row 1: `1 2 3 0 1`
   - Row 2: `0 1 2 3 1`
   - Row 3: `1 0 1 2 2`
   - Row 4: `2 1 0 1 0`
   - Row 5: `1 2 1 0 1`
3. **Filter (3×3):**
   - Row 1: `1 0 -1`
   - Row 2: `1 0 -1`
   - Row 3: `1 0 -1`
4. **Chosen values:** stride = 1, padding = 0
5. **Output size:** (N + 2P − F)/S + 1 = (5 + 0 − 3)/1 + 1 = **3**, so the output is 3×3.
6. **Step-by-step:** The middle filter column is 0, so each patch row gives (left − right).
   - O(1,1) = (1−3) + (0−2) + (1−1) = **−4**
   - O(1,2) = (2−0) + (1−3) + (0−2) = **−2**
   - O(1,3) = (3−1) + (2−1) + (1−2) = **2**
   - O(2,1) = (0−2) + (1−1) + (2−0) = **0**
   - O(2,2) = (1−3) + (0−2) + (1−1) = **−4**
   - O(2,3) = (2−1) + (1−2) + (0−0) = **0**
   - O(3,1) = (1−1) + (2−0) + (1−1) = **2**
   - O(3,2) = (0−2) + (1−1) + (2−0) = **0**
   - O(3,3) = (1−2) + (0−0) + (1−1) = **−1**
7. **Output feature map (3×3):**
   - Row 1: `-4 -2 2`
   - Row 2: `0 -4 0`
   - Row 3: `2 0 -1`

# Part A: Load the data

````python
import kagglehub
import pandas as pd
import os

# Download dataset
path = kagglehub.dataset_download(
    "arashnic/faces-age-detection-dataset"
)

print("Dataset path:", path)

# Dynamically search for train.csv in case of nested folder structure
csv_path = None
if os.path.exists(os.path.join(path, "train.csv")):
    csv_path = os.path.join(path, "train.csv")
else:
    for root, dirs, files in os.walk(path):
        if "train.csv" in files:
            csv_path = os.path.join(root, "train.csv")
            break

if csv_path is None:
    raise FileNotFoundError("Could not locate train.csv inside the downloaded dataset.")

print("Found train.csv at:", csv_path)
df = pd.read_csv(csv_path)


print(df.head())
print("Dataset shape:", df.shape)
print("Columns:", df.columns.tolist())
````

# Part B: Preprocess the images

````python
import torch
from torch.utils.data import Dataset
from torchvision import transforms
from PIL import Image

label_map = {"Young": 0, "Middle": 1, "Old": 2}
df["label"] = df["Class"].map(label_map)


train_tf = transforms.Compose([
    transforms.Resize((64, 64)),
    transforms.RandomHorizontalFlip(),
    transforms.RandomRotation(10),
    transforms.ToTensor(),
    transforms.Normalize([0.5]*3, [0.5]*3),
])

val_tf = transforms.Compose([
    transforms.Resize((64, 64)),
    transforms.ToTensor(),
    transforms.Normalize([0.5]*3, [0.5]*3),
])

class FaceDataset(Dataset):
    def __init__(self, dataframe, folder, transform):
        self.df = dataframe.reset_index(drop=True)
        self.folder = folder
        self.transform = transform
    def __len__(self):
        return len(self.df)
    def __getitem__(self, idx):
        row = self.df.iloc[idx]
        img = Image.open(os.path.join(self.folder, row["ID"])).convert("RGB")
        return self.transform(img), int(row["label"])
````

# Part C: Split and batch

````python
from sklearn.model_selection import train_test_split
from torch.utils.data import DataLoader
import os

label_map = {
    "YOUNG": 0,
    "MIDDLE": 1,
    "OLD": 2
}

df["label"] = df["Class"].map(label_map)

img_dir = os.path.join(
    os.path.dirname(csv_path),
    "Train")

print("Image folder:", img_dir)

train_df, val_df = train_test_split(
    df,
    test_size=0.2,
    stratify=df["label"],
    random_state=42)


train_ds = FaceDataset(
    train_df,
    img_dir,
    train_tf)

val_ds = FaceDataset(
    val_df,
    img_dir,
    val_tf)

train_loader = DataLoader(
    train_ds,
    batch_size=64,
    shuffle=True,
    num_workers=2)

val_loader = DataLoader(
    val_ds,
    batch_size=64,
    shuffle=False,
    num_workers=2)

print("Train size:", len(train_ds))
print("Validation size:", len(val_ds))
````

# Part D: Build the CNN

````python
import torch
import torch.nn as nn

def block(in_c, out_c):
    return nn.Sequential(
        nn.Conv2d(in_c, out_c, 3, padding=1),
        nn.BatchNorm2d(out_c),
        nn.ReLU(),

        nn.Conv2d(out_c, out_c, 3, padding=1),
        nn.BatchNorm2d(out_c),
        nn.ReLU(),

        nn.MaxPool2d(2, 2)
    )


# CNN Model
class AgeCNN(nn.Module):

    def __init__(self):
        super().__init__()

        self.features = nn.Sequential(
            block(3, 32),
            block(32, 64),
            block(64, 128),
            block(128, 256) )

        self.classifier = nn.Sequential(
            nn.Flatten(),

            nn.Linear(256 * 4 * 4, 512),
            nn.ReLU(),
            nn.Dropout(0.5),

            nn.Linear(512, 128),
            nn.ReLU(),
            nn.Dropout(0.3),

            nn.Linear(128, 3)
        )

    def forward(self, x):
        x = self.features(x)
        x = self.classifier(x)
        return x

device = "cuda" if torch.cuda.is_available() else "cpu"

model = AgeCNN().to(device)

print(model)
print("Device:", device)
````

# Part E: Train and evaluate

````python
import numpy as np
import matplotlib.pyplot as plt

counts = train_df["label"].value_counts().sort_index().values

weights = torch.tensor(
    counts.sum() / (3 * counts),
    dtype=torch.float
).to(device)


criterion = nn.CrossEntropyLoss(weight=weights)

optimizer = torch.optim.Adam(
    model.parameters(),
    lr=0.001)


epochs = 15

history = {
    "tr_loss": [],
    "va_loss": [],
    "tr_acc": [],
    "va_acc": []}


for epoch in range(epochs):

    model.train()

    total_loss = 0
    correct = 0
    total = 0

    for x, y in train_loader:

        x = x.to(device)
        y = y.to(device)

        optimizer.zero_grad()

        output = model(x)

        loss = criterion(output, y)

        loss.backward()
        optimizer.step()

        total_loss += loss.item() * x.size(0)
        correct += (output.argmax(1) == y).sum().item()
        total += x.size(0)


    train_loss = total_loss / total
    train_acc = correct / total

    model.eval()

    val_loss = 0
    val_correct = 0
    val_total = 0

    with torch.no_grad():

        for x, y in val_loader:

            x = x.to(device)
            y = y.to(device)

            output = model(x)

            loss = criterion(output, y)

            val_loss += loss.item() * x.size(0)
            val_correct += (output.argmax(1) == y).sum().item()
            val_total += x.size(0)


    val_loss = val_loss / val_total
    val_acc = val_correct / val_total

    history["tr_loss"].append(train_loss)
    history["va_loss"].append(val_loss)
    history["tr_acc"].append(train_acc)
    history["va_acc"].append(val_acc)


    print(f"Epoch {epoch+1}/{epochs} | "
        f"Train Loss: {train_loss:.4f} "
        f"Acc: {train_acc:.3f} | "
        f"Val Loss: {val_loss:.4f} "
        f"Acc: {val_acc:.3f}")
````

````python
plt.figure(figsize=(12, 4))

plt.subplot(1, 2, 1)

plt.plot(history["tr_loss"], label="Train")
plt.plot(history["va_loss"], label="Validation")

plt.title("Loss")
plt.xlabel("Epoch")
plt.ylabel("Loss")
plt.legend()

plt.subplot(1, 2, 2)

plt.plot(history["tr_acc"], label="Train")
plt.plot(history["va_acc"], label="Validation")

plt.title("Accuracy")
plt.xlabel("Epoch")
plt.ylabel("Accuracy")
plt.legend()


plt.show()
````
